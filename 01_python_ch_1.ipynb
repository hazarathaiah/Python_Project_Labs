## **Calculate Market Capitalization 📈**

### **Concepts :** Functions, type hints, datatypes, Operations, Exception Handling & a first test

**Market Cap = Price per Share × Total Shares Outstanding**

It tells you the total market value of a company. We'll write a robust function
that handles bad inputs gracefully using `try/except`


### **1.1 The `calculate_market_cap` function**

Two things to notice in the code below:

- **Type hints** (`price_per_share: float | str`) - these don't change how the code runs; they document what goes in and what comes out.
- **Convert first, validate second.** Only the conversion can raise an exception, so only the conversion goes inside `try`.

```python
def calculate_market_cap(price_per_share: float | str,
                         total_shares: float | str) -> float | None:
    """Return price x shares, or None if either input is invalid.

    Args:
        price_per_share: Price of one share (Accepts a number or a numeric string)
        total_shares:    Shares outstanding. (Accepts a number or a numeric string)

    Returns:
        The market cap as a float, or None if the inputs could not be used.
    """
    try:
        price  = float(price_per_share)
        shares = float(total_shares)
    except (TypeError, ValueError) as e:
        print(f"Error calculating market cap: {e}")
        return None

    if price < 0 or shares < 0:
        print("Error: price and shares must be non-negative.")
        return None

    return price * shares
```

### **1.2 Try it out**

```python
# Example: RIL stock
price_per_share = 2850.50
total_shares    = "-1000"

market_cap = calculate_market_cap(price_per_share, total_shares)

if market_cap is None:
    print("Can't calculate market cap, check the inputs.")
else:
    print(f"Market Capitalization: ${market_cap:,.2f}")
```

### **1.3 Datatypes you'll actually use**

So far we've used `float` and `int`. Two more matter far more than either, because they're the shape almost all real data arrives in:

1. list
2. dict

Combined as **`list[dict]`** : a list of dictionaries —> they describe a table, a CSV file, a JSON API response, and a chat message history.

```python
# A list of dicts — the shape of a CSV file, a table, or a JSON API response.
portfolio: list[dict] = [
    {"ticker": "TCS",  "price": 2850.50, "shares": 1_000_000},
    {"ticker": "TSLA", "price":  245.30, "shares": 3_185_000_000},
    {"ticker": "IBM", "price":  380.25, "shares": 7_430_000_000},
]

# Loop over the list, work on one dict at a time.
for row in portfolio:
    row["market_cap"] = calculate_market_cap(row["price"], row["shares"])

print(f"{'Ticker':<20} {'Market Cap':>8}\n")
for row in portfolio:
    print(f"{row['ticker']:<8} {row['market_cap']:>22,.2f}")
```
