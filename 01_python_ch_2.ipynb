### Concepts

### **3.1 The `RatioCalculator` class**

```python
import csv, io
from typing import Dict, List


class RatioCalculator:
    """Reads stock data and computes financial ratios."""

    def __init__(self, raw_csv: str) -> None:
        self.raw_csv = raw_csv

    @staticmethod
    def safe_ratio(numerator: str, denominator: str) -> float | str:
        """Return numerator/denominator as float, or '' if invalid."""
        try:
            num = float(numerator)
            den = float(denominator)
            if den == 0:
                return ""
            return round(num / den, 6)
        except (TypeError, ValueError):
            return ""

    def calculate_row(self, row: Dict[str, str]) -> Dict[str, str]:
        row["price_to_book_ratio"] = self.safe_ratio(row.get("price", ""), row.get("book_value_per_share", ""))
        row["pe_ratio"]            = self.safe_ratio(row.get("price", ""), row.get("eps", ""))
        return row

    def run(self) -> List[Dict]:
        reader = csv.DictReader(io.StringIO(self.raw_csv))
        return [self.calculate_row(row) for row in reader]
```

### **3.2 Use the class**

```python
from google.colab import files

# Upload in.csv
uploaded = files.upload()
```

```python
import io

filename = list(uploaded.keys())[0]
CSV_DATA = uploaded[filename].decode("utf-8")

print(f"✅ Loaded '{filename}' for Chapter 3")
```

```python
calc    = RatioCalculator(CSV_DATA)
results = calc.run()

for r in results:
    print(f"{r['ticker']:<8}  P/B={r['price_to_book_ratio']:>12}  P/E={r['pe_ratio']:>12}")
```

### **3.3 Key differences vs Chapter 2**

| Procedural (Ch 2) | OOP (Ch 3) |
|---|---|
| `safe_ratio()` is a free function | `safe_ratio()` is a `@staticmethod` on the class |
| Logic spread across functions | Bundled inside `RatioCalculator` |
| Hard to reuse / test in isolation | Easy to instantiate with different data |
